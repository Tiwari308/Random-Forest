```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df['YearsExperience']=le.fit_transform(df['YearsExperience'])
```

```python
df.YearsExperience.value_counts()
```

```python
sns.countplot(x='Salary',data=df)
```

```python
plt.figure(figsize=(20,18))
pd.crosstab(df.Salary,df.YearsExperience).plot(kind="bar")
```

```python
df.columns
```

```python
X1=df[['YearsExperience']]
y1=df['Salary']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X1, y1, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
y_predict=model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_predict)
print(cm)
```

```python
from sklearn.metrics import classification_report
clr=classification_report(y_test,y_predict)
print(clr)
```

```python
sns.heatmap(cm,annot=True)
```

```python
from sklearn.ensemble import BaggingClassifier
```
